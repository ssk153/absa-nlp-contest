```python
import pandas as pd
import os

data_candidates = ["../data/contest2_train.csv", "data/contest2_train.csv", "contest2_train.csv"]
data_path = next((p for p in data_candidates if os.path.exists(p)), "data/contest2_train.csv")
df = pd.read_csv(data_path)

print("Shape:", df.shape)
print()

print("Columns:")
print(df.columns.tolist())
print()

print("Missing values:")
print(df.isna().sum())
print()

print("Unique IDs:", df["id"].nunique())
print()

print("Aspect distribution:")
print(df["aspectCategory"].value_counts())
print()

print("Polarity distribution:")
print(df["polarity"].value_counts())
print()

print("Aspect × polarity:")
print(
    pd.crosstab(
        df["aspectCategory"],
        df["polarity"]
    )
)
```
